# Initial Exploratory Data Analysis

This notebook performs an initial, read-only exploration of `data/raw/car_prices_tr.csv`. It does not clean, modify, preprocess, or save the dataset.

## 1. Import pandas

```python
from pathlib import Path

import pandas as pd
```

## 2. Load the raw dataset

The path works when the notebook is opened from either the repository root or the `notebooks` directory.

```python
data_path = Path("data/raw/car_prices_tr.csv")
if not data_path.exists():
    data_path = Path("../data/raw/car_prices_tr.csv")

df = pd.read_csv(data_path)
```

## 10. Unique values in categorical columns

Categorical columns are identified from pandas `object`, `category`, and `string` data types. Missing values are included as a distinct value in the counts.

```python
categorical_columns = df.select_dtypes(
    include=["object", "category", "string"]
).columns

unique_counts = (
    df[categorical_columns]
    .nunique(dropna=False)
    .rename("unique_values")
    .to_frame()
)
unique_counts
```

```python
import matplotlib.pyplot as plt

top_brands = df["marka"].value_counts().head(10)

plt.figure(figsize=(10,5))
ax = top_brands.plot(kind="bar")

plt.title("Top 10 Car Brands")
plt.xlabel("Brand")
plt.ylabel("Number of Listings")

for i, value in enumerate(top_brands):
    ax.text(i, value + 15, str(value), ha="center")

plt.show()
```
