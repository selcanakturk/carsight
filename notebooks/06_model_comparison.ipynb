# Model Comparison

In this notebook, we compare the performance of the baseline Linear Regression model and the Random Forest Regressor.

Evaluation metrics:

- MAE
- RMSE
- R² Score

The goal is to determine which model performs better for second-hand car price prediction.

```python
import pandas as pd

import matplotlib.pyplot as plt
```

```python
results = pd.DataFrame(
    {
        "Model": [
            "Linear Regression",
            "Random Forest",
        ],
        "MAE": [
            358262,
            157289,
        ],
        "RMSE": [
            738093,
            556297,
        ],
        "R²": [
            0.236,
            0.566,
        ],
    }
)

results
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    results["Model"],
    results["R²"]
)

plt.title("R² Score Comparison")

plt.ylabel("R²")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    results["Model"],
    results["MAE"]
)

plt.title("MAE Comparison")

plt.ylabel("TL")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    results["Model"],
    results["RMSE"]
)

plt.title("RMSE Comparison")

plt.ylabel("TL")

plt.show()
```

# Conclusion

Random Forest outperformed Linear Regression on every evaluation metric.

Compared to the baseline model, it:

- reduced MAE considerably,
- reduced RMSE,
- increased the R² score from 0.236 to 0.566.

This indicates that Random Forest captures nonlinear relationships in second-hand car pricing much better than Linear Regression.
