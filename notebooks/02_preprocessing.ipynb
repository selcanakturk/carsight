```python
from pathlib import Path
import pandas as pd

data_path = Path("data/raw/car_prices_tr.csv")

if not data_path.exists():
    data_path = Path("../data/raw/car_prices_tr.csv")

df = pd.read_csv(data_path)
```

```python
df.head()
```

```python
df[["kilometre(Km)", "yıl"]].head(10)
```

```python
df["gercek_yil"] = df["kilometre(Km)"]
df["gercek_km"] = df["yıl"]
```

```python
df[["gercek_yil", "gercek_km"]].head(10)
```

```python
df["kilometre(Km)"], df["yıl"] = df["yıl"], df["kilometre(Km)"]
```

```python
df[["kilometre(Km)", "yıl"]].head(10)
```

```python
df[["kilometre(Km)", "yıl", "fiyat(TRY)"]].dtypes
```

```python
df["kilometre(Km)"] = (
    df["kilometre(Km)"]
        .str.replace(".", "", regex=False)
        .str.replace(" km", "", regex=False)
)

df["kilometre(Km)"] = pd.to_numeric(
    df["kilometre(Km)"],
    errors="coerce"
)
```

```python
df[["kilometre(Km)", "yıl"]].head()
```

```python
df[["kilometre(Km)", "yıl"]].dtypes
```

```python
df["fiyat(TRY)"].head(10)
```

```python
df["fiyat(TRY)"] = (
    df["fiyat(TRY)"]
        .str.replace(".", "", regex=False)
        .str.replace(" TL", "", regex=False)
)

df["fiyat(TRY)"] = pd.to_numeric(
    df["fiyat(TRY)"],
    errors="coerce"
)
```
